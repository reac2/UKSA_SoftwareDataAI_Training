```python
# A lambda function is known as an anonymous function as it is not bound to an identifier,
# A regular function (defined with def) can be reused many times. A lambda function are single-use functions that can be inlined. 
# Since you only use them once, 
# they don't need an identifier

```

```python
# If we wanted to write a reusable function to double a number:

def doubleNum(num):
    return 2 * num

# As a single use lambda expression

# This is known as an antipattern as lambda functions are not supposed to be assigned to an identifier - it defeats the point of an anonymous function!
doubleNum = lambda num : 2 * num # This takes the identifier doubleNum from the namespace which is also bad practice with lambda functions.


print(doubleNum(5))

```

```python
# You can create custom sorts
# With a selection sort function
def sortByGrade(studentGradeList):
   for i in range(len(studentGradeList)):
       smallestGradeStudent = i
       for j in range(i,len(studentGradeList)):
           if studentGradeList[j][1] < studentGradeList[smallestGradeStudent][1]:
               smallestGradeStudent = j
       studentGradeList[smallestGradeStudent], studentGradeList[i] = studentGradeList[i], studentGradeList[smallestGradeStudent]
               
   return studentGradeList 
        
studentGradeList = [["Aaron",56],["Beth",15],["Carrie",80], ["Daniel",12]]

print(f"With a function: {sortByGrade(studentGradeList)}")
print(f"With a lambda function {sorted(studentGradeList, key=lambda grade: grade[1])}")
```

As we can see, using the lamda function to make this custom sort was much easier, quicker, and it will be more efficient.
